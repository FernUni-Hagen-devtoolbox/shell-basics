### Aufgabe 1 - Wer bin ich und wo bin ich?
Mit dem Befehl `whoami` können Sie sehen, mit welchem Benutzer Sie aktuell eingeloggt sind. Geben Sie den Befehl in die nachfolgende Code-Zeile ein und führen Sie ihn aus, indem Sie `Strg+Enter` drücken.

```bash
whoami
```

In der Ausgabe erscheint der Benutzername. Als nächstes prüfen Sie durch den Befehl `pwd` in welchem Verzeichnis Sie sich gerade befinden. `pwd` steht dabei für *print working directory*.

```bash
pwd
```

Sie befinden sich in Ihrem home-Verzeichnis. Wir werden im späteren Verlauf des Moduls noch mehr über dieses Verzeichnis und seine Eigenschaften lernen.

Bevor wir das Verzeichnis wechseln, schauen wir uns den Inhalt des aktuellen Verzeichnisses an. Dazu gibt es den Befehl `ls` (für *list*).

```bash
ls
```

Um das Verzeichnis anschließend zu wechseln, gibt es den Befehl `cd`. Dabei steht `cd` für *change directory*. `cd` nimmt als Argument den Pfad des Ordnerns, zu dem Sie wechseln wollen. Wechseln Sie nun in den test-Ordner.

```bash
cd test
```

Überprüfen Sie mit pwd, ob der Wechsel geklappt hat.

```bash
pwd
```

In der Ausgabe sehen Sie, dass der test-Ordner Teil Ihres home-Verzeichnisses ist. Da Sie sich vor dem `cd` Befehl in diesem befunden haben, hat es gereicht für den Wechsel den **relativen Pfad** zu verwenden.

Der Output von `pwd` gibt immer den **absoluten Pfad** aus. Das ist der gesamte Pfad von der Wurzel des Dateisystems aus. Diese Wurzel wird auf Englisch passenderweise **root** genannt und wird durch einen Schrägstrich **/** repräsentiert.
Wechseln Sie im nächsten Block per `cd` zum root-Verzeichnis.

```bash
cd /
```

Sie befinden sich jetzt im root-Verzeichnis. Wenn Sie nun zurück in den test-Ordner wechseln wollen, können Sie nicht denselben relativen Pfad wie oben verwenden. Sie müssen nun den relativen Pfad von / aus nutzen.

```bash
cd home/jovyan/test
```

```bash
pwd
```

Dieser relative Pfad unterscheidet sich nur durch den fehlenden Schrägstrich vom absoluten Pfad. 
Da das home-Verzeichnis einer der wichtigsten Orte für einen/eine Nutzer:in ist und dementsprechend häufig verwendet wird, gibt es eine syntaktische Abkürzung dafür, das Tilde-Symbol **~**.

Um zu überprüfen, wofür das Symbol steht, können Sie den `echo` Befehl nutzen. `echo` gibt einfach das als Argument auf dem Bildschirm aus, was übergeben wurde. Handelt es sich dabei um Variablen oder andere Platzhalter, werden diese vorher aufgelöst. So auch hier, die Shell wandelt die Tilde vorher in den absoluten Pfad um. Probieren Sie es aus!

```bash
echo ~
```

Wie Sie sehen, ist das Tilde-Symbol ein Platzhalter für den absoluten Pfad des home-Verzeichnisses. Da das Wechseln in dieses Verzeichnis so regelmäßig geschieht, gibt es eine noch einfachere Variante dafür. Sie können einfach den `cd`-Befehl ohne Argument verwenden und gelangen dann zum home-Verzeichnis.

Um in das home-Verzeichnis zu wechseln, haben Sie also immer vier Möglichkeiten:
- Sie nutzen den relativen Pfad, vom aktuellen Pfad aus
- Sie nutzen den absoluten Pfad `cd /home/<Benutzer:in>`
- Sie nutzen das Tilde-Symbol `cd ~`
- Sie nutzen nur `cd`


```bash
cd /
```

```bash
cd
```

```bash
pwd
```

Das Tilde-Symbol wird vor allem in solchen Fällen benutzt, in denen Sie in einen Unterordner des home-Verzeichnisses wechseln wollen. Dann können Sie den ersten Teil des absoluten Pfads dadurch ersetzen.
Gehen Sie nun nochmal zurück in den root-Ordner und wechseln Sie zum test-Ordner, der sich in Ihrem home-Verzeichnis befindet. Nutzen Sie dafür den absoluten Pfad mit Tilde-Symbol.

```bash
cd /
```

```bash
cd ~/test
```

Neben der Tilde gibt es noch weitere Platzhalter für das relative Navigieren. Wir nutzen wieder `ls` um uns den Inhalt eines Ordnerns anzusehen. Allerdings reicht der einfache Befehl nicht aus, denn die Platzhalter sind standardmäßig versteckt. Um diese sichtbar zu machen, müssen wir uns das Verzeichnis mit der Option `-a` ansehen. Option `-a` steht für 'all' und listet auch alle versteckten Dateien (und Ordner) auf, die mit einem Punkt beginnen.
    
Führen Sie `ls -a` aus. Sie werden dort auch einen simplen Punkt **.** und einen Doppelpunkt **..** sehen. Der einzelne Punkt repräsentiert das aktuelle Verzeichnis, in dem Sie sich gerade befinden. Der Doppelpunkt **..** repräsentiert das relative Elternverzeichnis.

```bash
ls -a
```

Wechseln Sie nun zurück in Ihr home-Verzeichnis (das Elternverzeichnis des test-Ordners), indem Sie den doppelten Punkt verwenden.

```bash
cd ..
```

```bash
pwd
```
