## Shell Basics - Lernpfad: Anfänger:innen - Übung 02

### Aufgabe 1 - Verzeichnisse und Dateien erstellen
In dieser Übung werden wir Dateien und Verzeichnisse verwalten. Zuerst legen wir mit dem Befehl `mkdir` (make directory) ein neues Verzeichnis an. Führen Sie den folgenden Befehl aus.

```bash
mkdir meine_dateien
```

Überprüfen Sie nun, ob das neue Verzeichnis erzeugt wurde.

```bash
ls
```

Mit dem Befehl `touch` können Sie eine neue, leere Datei erstellen. Wir legen nun eine Textdatei in unserem neuen Ordner an. Wir nutzen dafür den relativen Pfad.

```bash
touch meine_dateien/notiz.txt
```

Wir überprüfen, ob die Datei erzeugt wurde wieder mit `ls`. Diesmal nutzen wir den Ordner als Argument.

```bash
ls meine_dateien
```

### Aufgabe 2 - Dateien lesen und kopieren
Um den Inhalt einer Datei anzeigen zu lassen, verwenden wir `cat`. Dabei steht `cat` für *concatenate*. Der Befehl ist also eigentlich dafür gedacht Ausgaben miteinander zu verknüpfen, hat sich aber entgegen dieser Funktion zur Ausgabe von Dateien etabliert. Da unsere Textdatei leer ist, fällt die Ausgabe entsprechend aus. Prüfen Sie zusätzlich die Datei *wichtige_datei* im home-Verzeichnis.

```bash
cat meine_dateien/notiz.txt
```

```bash
cat wichtige_datei
```

Nun möchten wir diese Datei kopieren. Dazu dient der Befehl `cp` (copy). Der Befehl `cp` erwartet zuerst die Quelldatei, dann das Ziel.

```bash
cp meine_dateien/notiz.txt meine_dateien/notiz_kopie.txt
```

```bash
ls meine_dateien
```

Die in der ersten Übung gelernten Platzhalter wie `~` (home-Verzeichnis) und `.` (aktuelles Verzeichnis) lassen sich hervorragend mit `cp` kombinieren. Kopieren wir zum Beispiel die `wichtige_datei` unter Verwendung der Platzhalter in den Ordner `meine_dateien` (der Punkt `.` verweist dabei auf den aktuellen Pfad, in dem wir uns befinden, danach folgt der neue Dateiname):

```bash
cp ~/wichtige_datei ./meine_dateien/wichtige_datei
```

```bash
ls meine_dateien
```

### Aufgabe 3 - Verschieben und Umbenennen
Mit dem Befehl `mv` (move) lassen sich Dateien und Verzeichnisse verschieben. Dieser Befehl wird auch verwendet, um Dateien einfach nur umzubenennen (indem Sie sie in das gleiche Verzeichnis unter einem anderen Namen verschieben). Analog zu `cp` erwartet `mv` auch wieder zuerst die Quelle und dann das Ziel als Argument.

```bash
mv meine_dateien/notiz_kopie.txt meine_dateien/wichtig.txt
```

```bash
ls meine_dateien
```

Auch der Platzhalter `..` (Elternverzeichnis) ist als Ziel für `mv` oder `cp` sehr nützlich. Wechseln wir kurz in den Ordner `meine_dateien` und verschieben die eben umbenannte Datei `wichtig.txt` eine Ebene nach oben in das home-Verzeichnis. Anschließend wechseln wir wieder zurück aus unserem Unterordner:

```bash
cd meine_dateien
```

```bash
mv wichtig.txt ../wichtig.txt
```

```bash
cd ..
```

```bash
ls
```

**Hinweis: Zieldateinamen weglassen (vereinfachte Schreibweise)**

Wenn Sie eine Datei mit `cp` oder `mv` in ein anderes Verzeichnis befördern und den ursprünglichen Namen beibehalten möchten, reicht es aus, als Ziel **nur das Verzeichnis** anzugeben. Sie müssen den Dateinamen im Zielpfad nicht zwingend nochmals tippen!

Kopieren Sie die eben verschobene `wichtig.txt` mit `cp` zurück in den Ordner und verschieben Sie sie danach mit `mv` wieder eine Ebene höher (`..`) und geben Sie diesmal als Ziel jedes Mal nur den Ordnernamen bzw. den Platzhalter an:

```bash
cp wichtig.txt meine_dateien/
```

```bash
cd meine_dateien
```

```bash
mv wichtig.txt ..
```

```bash
cd ..
```

```bash
ls
```

### Aufgabe 4 - Löschen mit Vorsicht!
Um Dateien zu löschen, gibt es den Befehl `rm` (remove).
Bei Verzeichnissen benutzt man `rmdir` (remove directory). Das funktioniert allerdings nur, *wenn das Verzeichnis komplett leer ist.*

**Achtung:** In der Shell gibt es standardmäßig keinen Papierkorb. Was gelöscht ist, ist unwiderruflich weg! Passen Sie daher mit `rm` gut auf.

```bash
rm meine_dateien/notiz.txt
```

```bash
rm meine_dateien/wichtige_datei
```

```bash
rm wichtig.txt
```

```bash
ls meine_dateien
```

Da das Verzeichnis nun wieder leer ist, können wir es abschließend gefahrlos mit `rmdir` löschen.

```bash
rmdir meine_dateien
```

```bash
ls
```

### Aufgabe 5 - Nicht leere Verzeichnisse löschen
Was ist, wenn wir ein Verzeichnis löschen wollen, das noch nicht leer ist? Mit `rmdir` erhalten Sie in einem solchen Fall eine Fehlermeldung. Hier hilft stattdessen der Befehl `rm` weiter, wenn wir ihn mit der Option `-r` (für *recursive* / rekursiv) kombinieren. Dadurch werden der Ordner und **alle** darin enthaltenen Verzeichnisse und Dateien gelöscht.

Lassen Sie uns kurz einen neuen Ordner mitsamt einer Datei erstellen und diesen sofort wieder rekursiv löschen:

```bash
mkdir voller_ordner
```

```bash
touch voller_ordner/beispiel.txt
```

```bash
rm -r voller_ordner
```

```bash
ls
```
