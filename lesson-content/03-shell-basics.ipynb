### Aufgabe 1 - Schnelle Hilfe zur Selbsthilfe
Immer wieder werden Sie Befehle nutzen wollen, deren genaue Optionen Sie vergessen haben. Viele Programme bieten dafür eine Kurzanleitung über die Option `--help` an.

Rufen Sie die Hilfe für den bekannten Befehl `ls` im Terminal auf der rechten Seite auf. Scrollen Sie dort durch die Ausgabe.

```bash
ls --help
```

Das ist eine Menge Text! Wenn Sie nach oben scrollen, sehen Sie eine lange Liste aller möglichen Optionen für `ls`. 

Suchen Sie in der Ausgabe nach der passenden Option, mit der Sie die Liste der Dateien "rückwärts" (reverse) sortieren können. Wenden Sie diesen Parameter nun auf den Ordner an, um dessen Inhalt in umgekehrter Reihenfolge anzeigen zu lassen:

<details>
  <summary>Lösung anzeigen</summary>

```bash
ls -r
```

</details>

### Aufgabe 2 - Den Überblick behalten mit `whatis`

Stellen Sie sich vor, Sie stoßen auf einen unbekannten Befehl. Bevor Sie ihn ausführen, sollten Sie sich informieren.
Der Befehl `whatis` durchsucht die installierten Handbuchseiten und gibt eine kurze Beschreibung eines dort dokumentierten Kommandos aus. Er analysiert keine beliebigen Skripte aus dem Internet. Solche Skripte sollten Sie lesen und ihre Herkunft prüfen, bevor Sie sie ausführen.

Finden Sie im Terminal heraus, was der Befehl `mkdir` macht:

```bash
whatis mkdir
```

*Hinweis:* Vielleicht fällt Ihnen in der Ausgabe auf, dass dort `mkdir (1)` steht. Die Zahl in den Klammern steht für die verschiedenen Kapitel der Handbücher. Was das genau bedeutet, schauen wir uns direkt in der nächsten Aufgabe (beim Handbücher-Befehl `man`) etwas genauer an!

### Aufgabe 3 - In die Tiefe gehen mit den Handbüchern (`man`)

Wenn Ihnen `--help` nicht detailliert genug ist, kann ein Blick in das Handbuch helfen. Sie rufen es auf, indem Sie `man` gefolgt von dem entsprechenden Befehl eingeben. Die *Manpages* enthalten ausführliche Beschreibungen von Optionen und häufig auch Beispiele.

Rufen wir das offizielle Handbuch für den Kopier-Befehl `cp` auf:

```bash
man cp
```

Sie landen nun in einem sogenannten *Pager*. Das ist ein Programm, um größere Texte komfortabel im Terminal lesen zu können. Sie navigieren mit den Pfeiltasten und beenden den Pager mit `q`. Hilfe zum Pager erhalten Sie mit `h`.

**Exkurs: System-Sektionen (Die Zahlen in den Klammern)**

Erinnern Sie sich an die Ausgabe von `whatis mkdir` in Aufgabe 2? Dort waren Zahlen wie `(1)`, `(1p)` oder `(2)` zu sehen. Da wir nun die Handbücher (`man`) kennen, schauen wir uns diese Zahlen genauer an. Sie stehen für die verschiedenen **Sektionen** (Kapitel) der System-Handbücher:

- **(1)**: Steht für die **allgemeinen Terminal-Befehle** für Anwender:innen. Das ist die Sektion, die für uns relevant ist! (Das `p` bei manchen steht für eine POSIX-Spezifikation - einen Standard für Betriebssysteme, den Sie aktuell ignorieren können)
- **(2) und (3)**: Stehen für System- oder Bibliotheksaufrufe. Diese Sektionen richten sich an Programmierer:innen, die Software für das Betriebssystem schreiben 

Wenn mehrere Handbuchseiten denselben Namen tragen, können Sie die gewünschte Sektion ausdrücklich angeben, zum Beispiel mit `man 1 printf`. Ohne diese Angabe zeigt `man` die erste passende Seite seiner Suchreihenfolge.

### Aufgabe 4 - Freie Übung: Einen unbekannten Befehl zähmen

In einem Forumsbeitrag wird der Befehl `wc` für die Analyse von Textdateien empfohlen.
Nutzen Sie nun Ihre neuen Fähigkeiten im Terminal, um herauszufinden, was dieser Befehl tut und wie man ihn richtig benutzt.

**Ihre Aufgaben:**
1. Finden Sie als erstes mit `whatis` heraus, wofür `wc` überhaupt steht und was es grob macht
2. Rufen Sie die Kurzanleitung `--help` oder das Handbuch (`man`) für `wc` auf
3. Finden Sie in der Dokumentation heraus, mit welcher Option Sie **ausschließlich die Anzahl der Zeilen** (lines) ausgeben können
4. Wir haben in Ihrem home-Verzeichnis eine geheimnisvolle Datei namens `die_antwort` abgelegt. Wenden Sie `wc` in Kombination mit Ihrer soeben gefundenen Option auf diese Datei an, um zu zählen, wie viele Zeilen sie besitzt

<details>
  <summary>Lösung anzeigen</summary>

```bash
wc -l die_antwort
```

</details>
