### Aufgabe 1 - Schnelle Hilfe zur Selbsthilfe
Immer wieder werden Sie Befehle nutzen wollen, deren genaue Optionen Sie einfach mal vergessen haben. Zum Glück hat fast jeder Befehl in der Shell seine eigene kleine Kurzanleitung an Bord. Diese können Sie abrufen, indem Sie dem Befehl die Option `--help` anhängen.

Lassen Sie uns versuchen, die Hilfe für unseren bekannten Befehl `ls` aufzurufen. Führen Sie die nachfolgende Zelle aus.

**Hinweis:** Klicken Sie anschließend unter der Ausgabe auf *View as **srcollable Element***, um den gesamten Output sichten zu können.

```bash
ls --help
```

Das ist eine Menge Text! Wenn Sie nach oben scrollen, sehen Sie eine lange Liste aller möglichen Optionen für `ls`. 

Suchen Sie in der Ausgabe nach der passenden Option, mit der Sie die Liste der Dateien "rückwärts" (reverse) sortieren können. Wenden Sie diesen Parameter nun auf den Ordner an, um dessen Inhalt in umgekehrter Reihenfolge anzeigen zu lassen:

```bash
ls -r
```

### Aufgabe 2 - Den Überblick behalten mit `whatis`

Stellen Sie sich vor, Sie finden im Internet einen Befehl oder ein Skript und wissen nicht genau, was es macht. Bevor Sie es blind in Ihrem Terminal ausführen, sollten Sie sich kurz informieren!
Hier hilft der Befehl `whatis`. Er ist wunderbar dafür geeignet die System-Handbücher zu durchsuchen und eine schnelle, einzeilige Kurzbeschreibung zu einem bestimmten Kommando auszugeben.

Lassen Sie uns in der nächsten Zelle herausfinden, was der Befehl `mkdir` eigentlich genau macht:

```bash
whatis mkdir
```

*Hinweis:* Vielleicht fällt Ihnen in der Ausgabe auf, dass dort mehrmals `mkdir` steht, ergänzt um verschiedene Zahlen und Buchstaben in Klammern, wie `mkdir (1)`, `mkdir (1p)` oder `mkdir (2)`. Diese stehen für die verschiedenen Kapitel der Handbücher. Was das genau bedeutet, schauen wir uns direkt in der nächsten Aufgabe (beim Handbücher-Befehl `man`) etwas genauer an!

### Aufgabe 3 - In die Tiefe gehen mit den Handbüchern (`man`)

Wenn Ihnen `--help` nicht detailliert genug ist, dann kann ein Blick in das "Handbuch" (das Manual) helfen. Sie rufen es auf, indem Sie `man` gefolgt von dem entsprechenden Befehl eingeben. Die *manpages* enthalten in der Regel alle existierenden Parameter, Erklärungen und viele Beispiele.

*Wichtiger Hinweis:* In einem echten Terminal öffnet sich hierbei ein interaktiver Lesemodus (ein sogenannter *Pager*). In diesem Jupyter Notebook wird das Handbuch stattdessen direkt als langer, fortlaufender Text in die Ausgabe geschrieben. Klicken Sie wieder auf *View as srcollable Element*, um die gesamte Ausgabe zu erhalten.

Rufen wir das offizielle Handbuch für den Kopier-Befehl `cp` auf:

```bash
man cp
```

**Exkurs: System-Sektionen (Die Zahlen in den Klammern)**

Erinnern Sie sich an die Ausgabe von `whatis mkdir` in Aufgabe 2? Dort waren Zahlen wie `(1)`, `(1p)` oder `(2)` zu sehen. Da wir nun die Handbücher (`man`) kennen, schauen wir uns diese Zahlen genauer an. Sie stehen für die verschiedenen **Sektionen** (Kapitel) der System-Handbücher:

- **(1)**: Steht für die **allgemeinen Terminal-Befehle** für Anwender:innen. Das ist die Sektion, die für uns relevant ist! (Das `p` bei manchen steht für eine POSIX-Spezifikation - einen Standard für Betriebssysteme, den Sie aktuell ignorieren können)
- **(2) und (3)**: Stehen für System- oder Bibliotheksaufrufe. Diese Sektionen richten sich an Programmierer:innen, die Software für das Betriebssystem schreiben 

Wenn Sie das Handbuch mit `man` zu einem Begriff aufrufen, wählt das System standardmäßig immer automatisch das Handbuch für den Terminal-Befehl aus Sektion (1) aus.
