### Aufgabe 1 - Benutzer und Gruppen

Jedes unixartige Betriebssystem basiert auf einer Rechteverwaltung. In den ersten Übungen haben wir bereits `whoami` genutzt, um unseren aktuellen Benutzernamen (`jovyan`) herauszufinden. 
Neben dem eigenen Benutzerkonto gehört man im System auch immer mindestens einer oder mehreren **Gruppen** an. Gruppen fassen verschiedene Nutzer:innen zusammen, um ihnen geteilte Rechte für bestimmte Dateien oder Dienste (wie z.B. Drucker oder das Ausführen von Systembefehlen) zu geben.

Finden Sie mit dem Befehl `groups` heraus, in welchen Gruppen Ihr aktueller Benutzer Mitglied ist:

```bash
groups
```

*Hinweis:* Welche Gruppen angezeigt werden, hängt von der Systemkonfiguration ab. In der aktuellen MyBinder-Umgebung gehört der Benutzer `jovyan` zur Gruppe `users`.

### Aufgabe 2 - Dateirechte anzeigen (`ls -l`)

Damit Sie sehen können, welche Rechte für eine Datei oder einen Ordner gesetzt sind, nutzen wir eine erweiterte Listenansicht von `ls`. Die Option `-l` (für *long format*) zeigt uns eine Menge an Metadaten an.

Lassen Sie uns eine neue Datei `geheim.txt` erstellen und dann `ls -l` darauf anwenden:

```bash
touch geheim.txt
```

```bash
ls -l geheim.txt
```

In der Ausgabe sehen Sie eine lange Zeile, zum Beispiel `-rw-r--r-- 1 jovyan users 0 Sep 30 14:00 geheim.txt`. Gruppe, Datum und Uhrzeit können in Ihrer Ausgabe abweichen.

**Der erste Teil (die Zehner-Kette)** verrät Ihnen die Dateirechte:
- Das erste Zeichen zeigt den Typ an (`-` für normale Datei, `d` für Verzeichnis)
- Danach folgen 3 Blöcke aus je 3 Zeichen für die Rechte von:
  1. der **besitzenden Person** der Datei (*User*, hier: jovyan)
  2. der zugehörigen **Gruppe** (*Group*, hier: users)
  3. allen **anderen** auf dem System (*Others*)

Die Buchstaben stehen für **r**ead (Lesen), **w**rite (Schreiben) und e**x**ecute (Ausführen). Ein `-` bedeutet, dass dieses Recht an der Stelle *nicht* gewährt ist.
`-rw-r--r--` bedeutet also: Die besitzende Person darf lesen und schreiben (`rw-`), die Gruppe darf nur lesen (`r--`) und alle anderen dürfen auch nur lesen (`r--`).

**Die weiteren Spalten** dieser Ausgabe bedeuten von links nach rechts:
- **1**: Die Anzahl der Verweise (Hardlinks) auf diese Datei. Was das genau ist, ignorieren wir an dieser Stelle
- **jovyan**: Die Person, der die Datei gehört
- **users**: Die zugehörige Gruppe der Datei
- **0**: Die Dateigröße in Bytes (hier 0, da wir die Datei gerade erst leer erstellt haben)
- **Sep 30 14:00**: Ein Beispiel für Datum und Uhrzeit der letzten Änderung
- **geheim.txt**: Der Dateiname


### Aufgabe 3 - Die Rechte verändern (`chmod`)

Wir wollen verhindern, dass außer uns noch irgendjemand die Datei `geheim.txt` lesen kann. Der Befehl zum Ändern der Rechte (Modus) lautet `chmod` (change mode).

*Hinweis zur Hierarchie:* Die Rechte einer Datei können normalerweise nur die Person ändern, der die Datei gehört, sowie der Systemadministrator `root`. Lese- oder Schreibrechte allein erlauben anderen Personen keine Änderung des Dateimodus.

Die Schreibweise für `chmod` ist simpel:
1. Für **wen** wollen wir das Recht ändern? `u` (user/Besitzer), `g` (group), `o` (others) oder `a` (all)
2. Wollen wir es **hinzufügen** (`+`) oder **entfernen** (`-`)?
3. **Welches** Recht? `r`, `w` oder `x`

Wir entziehen nun der Gruppe (`g`) und den anderen (`o`) das Leserecht (`r`), indem wir `go-r` verwenden:

```bash
chmod go-r geheim.txt
```

```bash
ls -l geheim.txt
```

Schauen Sie sich die neue Ausgabe an! Die Rechte für Gruppe und andere sind nun komplett auf `---` gesetzt. Nur Sie (`rw-`) haben noch Zugriff!

### Aufgabe 4 - Wir sperren uns selbst aus!

Rechte betreffen nicht nur andere Personen. Auch der besitzenden Person kann ein Recht entzogen werden. Was passiert wohl, wenn wir unserem eigenen Benutzer (`u`) das Leserecht (`r`) für die Datei wegnehmen und sie danach mit `cat` anzeigen wollen?

```bash
chmod u-r geheim.txt
```

```bash
cat geheim.txt
```

**Permission denied** Sie können die Datei nicht mehr lesen, obwohl Sie sie selbst erstellt haben. Geben wir uns das Recht mit `u+r` besser schnell wieder zurück:

```bash
chmod u+r geheim.txt
```

### Aufgabe 5 - Ausführbare Dateien (Skripte)

Wir haben Ihnen eine kleine Datei namens `zeit.sh` in das Verzeichnis gelegt. Solche Dateien mit der Endung `.sh` sind in der Regel kleine Programme, sogenannte Shell-Skripte. 
Versuchen wir einmal, dieses Programm auszuführen. Wir geben dafür `./zeit.sh` in die Konsole ein. 

**Exkurs: Warum das `./`?**
Wenn Sie einen Befehl wie `ls` eingeben, durchsucht die Shell die im Suchpfad `PATH` eingetragenen Verzeichnisse nach einem Programm mit diesem Namen. Ihr *aktuelles* Verzeichnis durchsucht sie dabei standardmäßig jedoch **nicht**. Das verhindert, dass Sie versehentlich ein Skript ausführen, das zufällig den gleichen Namen trägt wie ein gängiger Systembefehl.
Wenn Sie also ein Programm aus Ihrem aktuellen Verzeichnis starten wollen, müssen Sie der Shell explizit den Pfad dorthin mitgeben. Hierfür nutzen wir den zuvor gelernten Platzhalter `.` (für das aktuelle Verzeichnis). Da in UNIX-Dateipfaden Verzeichnisse und Dateinamen immer durch einen Schrägstrich (`/`) voneinander getrennt werden, hängen wir diesen direkt an den Punkt an. So entsteht die Kombination `./zeit.sh`.

```bash
./zeit.sh
```

Sie erhalten ein **Permission denied**! Das Programm wird blockiert. Warum?

Schauen wir uns die Rechte der Datei an:

```bash
ls -l zeit.sh
```

Sie sehen in der Rechteanzeige, dass dem Benutzer das entscheidende **x** (für e**x**ecute = Ausführen) fehlt. Das verhindert den direkten Aufruf über `./zeit.sh`. Es sagt nichts darüber aus, ob der Inhalt der Datei ungefährlich ist.

Fügen Sie Ihrem Benutzer (`u`) nun das Ausführen-Recht (`x`) für e**x**ecute hinzu und probieren Sie erneut, das Skript zu starten!

```bash
chmod u+x zeit.sh
```

```bash
./zeit.sh
```

Die aktuelle Uhrzeit wird ausgegeben. Durch das Ausführungsrecht ist der direkte Aufruf der Datei nun erlaubt.
